# Rosetta Docking: RAR + Curcumin
Receptor: `RAR.pdb` | Ligand: `Curcumina.sdf`

Run each cell in order.

```python
import os
os.chdir('/data')
print(os.listdir('.'))
```

## 0. Locate Rosetta tools
Finds `molfile_to_params.py` and the `docking_protocol` binary and stores their paths for later cells.

```python
import os
import glob

# Find molfile_to_params.py
molfile = None
for root in ['/Rosetta', '/usr/local', '/opt', '/home']:
    hits = glob.glob(f'{root}/**/molfile_to_params.py', recursive=True)
    if hits:
        molfile = hits[0]
        break
print('molfile_to_params.py:', molfile)

# Find docking_protocol binary
docking = None
which = os.popen('which docking_protocol 2>/dev/null').read().strip()
if which:
    docking = which
else:
    for root in ['/Rosetta', '/usr/local', '/opt']:
        hits = glob.glob(f'{root}/**/bin/docking_protocol*', recursive=True)
        if hits:
            docking = hits[0]
            break
print('docking_protocol:', docking)
```

## 1. Prepare Receptor
Removes hydrogens, water, alternate conformations, and non-standard HETATMs.

```python
with open('RAR.pdb') as f, open('RAR_clean.pdb', 'w') as out:
    for line in f:
        if line.startswith(('ATOM', 'HETATM')):
            name = line[12:16].strip()
            alt = line[16]
            resname = line[17:20].strip()
            if resname in ('HOH', 'WAT'):
                continue
            if line.startswith('ATOM') and name.startswith('H'):
                continue
            if alt not in (' ', 'A'):
                continue
            if line.startswith('HETATM') and resname not in ('SO4', 'GOL', 'EDO', 'PO4'):
                continue
            out.write(line)
        elif line.startswith(('TER', 'END')):
            out.write(line)
    out.write('END\n')

print('Lines in RAR_clean.pdb:', sum(1 for _ in open('RAR_clean.pdb')))
```

## 2. Generate Ligand Params
`molfile_to_params.py` reads the SDF directly and writes `CUR.params` + `CUR_000.pdb`.

```python
import os
assert molfile, 'Step 0 failed to find molfile_to_params.py'
os.system(f'python "{molfile}" Curcumina.sdf --name CUR')
import glob
print(sorted(glob.glob('CUR*')))
```

## 3. Build Starting Complex
Concatenates receptor + ligand pose into one PDB.

```python
!cat RAR_clean.pdb CUR_000.pdb > complex.pdb
!wc -l complex.pdb
```

## 4. Run Docking
RosettaLigand flexible docking. `-partners L_A` assumes the ligand is chain L in `CUR_000.pdb` (check the generated file if it fails).

```python
import os
assert docking, 'Step 0 failed to find docking_protocol'
os.system(f'"{docking}" '
          '-in:file:fullatom '
          '-in:file:receptor RAR_clean.pdb '
          '-extra_res_fa CUR.params '
          '-in:file:native complex.pdb '
          '-partners L_A '
          '-dock_pert 8 20 '
          '-ex1 -ex2 '
          '-nstruct 10 '
          '-out:suffix _docked '
          '-out:file:silent docked.out '
          '-run:constant_seed')
```

## 5. Analyze Results
Shows the 5 lowest-scoring structures.

```python
import os
if os.path.exists('docked.out'):
    os.system('grep "^SCORE:" docked.out | sort -k2 -n | head -5')
else:
    import glob
    print('No silent file. Output structures:', glob.glob('*_docked*.pdb'))
```
